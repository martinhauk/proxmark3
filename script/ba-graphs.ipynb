#### Functions

```python
def extract_vals(str, max):
    str_lines = str.split("\n")
    int_lines = [[int(x.removeprefix("0").removeprefix("0")) for x in line.split(",")] for line in str_lines]
    res = []
    for dist in range(0, int_lines[0][0]):
        res.append(100)
    for dist in int_lines[1]:
        res.append(dist)
    for dist in range(int_lines[0][len(int_lines[0]) - 1], max):
        res.append(0)
    return res

def extract_highest(str):
    str_lines = str.split("\n")
    int_lines = [[int(x.removeprefix("0").removeprefix("0")) for x in line.split(",")] for line in str_lines]
    highest = 0
    i = 0
    for dist in int_lines[1]:
        if(int_lines[1][i] != 0):
            highest = int_lines[0][i]
        i = i + 1
    return highest
```

```python
def line_comparison_plot(org_lines, adpt_lines, org_colors, adpt_colors, name, filename): 
    def draw_graph(ax, line, color):
        # print(line)
        d,m = line.split("\n")
        ds = [int(x) for x in d.split(",")]
        ms = [int(x) for x in m.split(",")]
        t = ax.plot(ds, ms, marker="o", color=color)
           
    # print(org_lines)
    # print(adpt_lines)
    fig, ax = plt.subplots()
    for i in range(int(org_lines[0][0]), len(org_lines)):
        org_line = org_lines[i]
        org_color = org_colors[i]
        draw_graph(ax, org_line, org_color)
        adpt_line = adpt_lines[i]
        adpt_color = adpt_colors[i]
        draw_graph(ax, adpt_line, adpt_color)
    
    ax.set_xlabel('Distanz zwischen Lesegerät und Tag')
    ax.set_ylabel('\nWinkel von Tag und Leseantenne zueinander',)
    ax.set_title(name)
    plt.tight_layout()
    plt.savefig(filename + ".svg", bbox_inches='tight')
    plt.show()
```

```python
def bar_comparison_plot(org, adpt, labels, name, filename):
    fig, ax = plt.subplots()
    x_axis = np.arange(len(labels))
    ax.bar(x_axis - 0.2, org, width=0.35,label="Originale Software")
    ax.bar(x_axis + 0.2, adpt, width=0.35,label="Angepasste Software")
    
    plt.xticks(x_axis, labels)
    plt.xlabel("Hindernisse") 
    plt.ylabel("Erreichte Distanz") 
    # plt.title(name) 
    plt.legend(loc="lower left") 
    plt.savefig(filename + ".svg", bbox_inches='tight')
    plt.show()
```

```python
def polar_comparison_plot(org, adpt, name, max_dist, filename):
    r = 100
    radius = [x for x in range(0,max_dist+1,1)]
    winkel = np.radians(range(90, 195, 15))
    winkelreverse = np.radians(range(0,105, 15))[::-1]
    # print(radius)
    # print(winkel)
    # print(winkelreverse)
    r, w = np.meshgrid(winkel, radius)
    rr, wr = np.meshgrid(winkelreverse, radius)
    # print(r)
    # print(w)

    fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
    ax.set_thetamin(0)
    ax.set_thetamax(180)
    ax.set_rorigin(-5)

    c = ax.contourf(r, w, org.T)
    ax.contourf(rr, wr, adpt.T)
    fig.colorbar(c, ax=ax, shrink=0.5,label='Erfolgreiche Messungen aus 100')
    labels = range(0, 195,15)

    ax.set_xlabel('Distanz zwischen Lesegerät und Tag', labelpad=0, verticalalignment="baseline")
    ax.xaxis.set_label_coords(0.475, 0.15)
    ax.set_xticks(np.radians(labels))
    ax.set_xticklabels([(str(90 - x ) if (x < 90) else str(x - 90)) + "°" for x in labels])

    # ax.set_title(name, y=1.02, pad=-14)
    
    ax.set_ylabel(
        '\nWinkel von Tag und Leseantenne zueinander\nOriginale Software                             Angepasste Software',
        rotation="horizontal")
    ax.yaxis.set_label_coords(0.52, 0.85)
    ax.yaxis.set_tick_params(label2On=True)
    # plt.gcf().set_size_inches(5, 4)
    # plt.savefig('charts/fig_size.png', dpi=200)
    plt.tight_layout()
    plt.savefig(filename + ".svg", bbox_inches='tight')
    plt.show()
```

```python
def print_tab(org, adpt,labels):
    angles = np.arange(0, 91, 15)
    
    print(tab_start)
    i = 0
    for line in org:
        lines = line.split("\n")
        distances = lines[0]
        percent = lines[1]
        tab_line = ( "        \\multirow{14}{*}{original} " if i == 0 else "") + """
        & \\multirow{2}{*}{""" + labels[i] + """} & mm & """ + distances + """ \\\\
        & & \\% & """ + percent + """ \\\\
        \\cline{2-4}""" + ( "\n\n        \\hline\n        \\hline\n" if i == 6 else "")
        print(tab_line)
        i = i + 1
    i = 0
    for line in adpt:
        lines = line.split("\n")
        angle = str(angles[i]) + "°"
        distances = lines[0]
        percent = lines[1]
        tab_line = ( "        \\multirow{14}{*}{angepasst} " if i == 0 else "") + """
        & \\multirow{2}{*}{""" + labels[i] + """} & mm & """ + distances + """ \\\\
        & & \\% & """ + percent + """ \\\\
        \\cline{2-4}"""
        print(tab_line)
        i = i + 1

    print(tab_end)
```
